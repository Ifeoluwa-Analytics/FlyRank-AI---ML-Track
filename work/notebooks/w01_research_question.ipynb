## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Research Question: Which visible content items are at the highest risk of sustained traffic loss, and how can we rank them into a review queue accompanied by human-readable reason codes to guide SEO content updates?

Decision: My work decides which decaying or underperforming pages should be placed into an editorial review queue for content updates, rewrites, or metadata optimization.

Who Acts on It: SEO strategists, content marketers, and editorial teams responsible for site maintenance.

Cost of a Wrong Call:
The cost of a wrong call can either be a False Positive(flagging a healthy page) or a False Negative(missing a decaying page).  

a) False Positive (Flagging a healthy page): Wasting of limited editorial capacity and budget rewriting content that did not require updating.

b) False Negative (Missing a decaying page): Can lead to an unmitigated loss of organic traffic, impressions, and search rankings to competitors over time.

```python
import os, sys, subprocess
import pandas as pd, numpy as np

# 1. Environment & Repository Setup
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
else:
    # find the repo root from wherever this kernel started
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")


# 2. Load starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 3. Filter population noise (keep visible pages with sufficient history)
# Filtering rules: impressions_90d > 0 and content_age_days >= 90
df_filtered = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].copy()

# 4. Define target label: content decay (trend_direction == 'down')
df_filtered["is_declining"] = df_filtered["trend_direction"] == "down"

# 5. Statistical metrics for Section 2 decision & capacity analysis
total_filtered = len(df_filtered)
total_declining = df_filtered["is_declining"].sum()
decline_rate = df_filtered["is_declining"].mean()
batch_size = 50  # Evaluates Top-50 editorial review capacity

print("=" * 55)
print("SECTION 2: DECISION, CAPACITY & COST ANALYSIS")
print("=" * 55)
print(f"Total Filtered Inventory        : {total_filtered:,} pages")
print(f"Declining Pages ('down' label)  : {total_declining:,} pages ({decline_rate:.1%})")

print("\n--- 1. Editorial Action & Queue Capacity ---")
print(f"• Batch Capacity Limit          : {batch_size} pages per review cycle")
print(f"• Batch Inventory Coverage       : {batch_size / total_filtered:.2%} of total pages")
print(f"• Batch Capture of Declining    : {batch_size / total_declining:.2%} of all declining candidates")

print("\n--- 2. Cost Baseline (False Positives vs. False Negatives) ---")
expected_random_positives = round(batch_size * decline_rate)
expected_random_fp = batch_size - expected_random_positives

print(f"• Random Selection Precision@50 : {decline_rate:.1%}")
print(f"• Expected Correct Calls (TP)   : ~{expected_random_positives} out of 50 pages")
print(f"• Expected Wasted Audits (FP)   : ~{expected_random_fp} out of 50 healthy pages")
print("=" * 55)
```
